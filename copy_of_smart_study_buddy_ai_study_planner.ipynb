```python
print("🤖 AI Smart Study Buddy")
print("I will help you study better!")

subject = input("Enter your subject: ")
difficulty = input("Is it Easy, Medium, or Hard? ")

if difficulty.lower() == "easy":
    print("Revise your", subject, "for 30 minutes.")
elif difficulty.lower() == "medium":
    print("Study your", subject, "for 1 hour and practice questions.")
elif difficulty.lower() == "hard":
    print("Study your", subject, "for 2 hours with short breaks.")
else:
    print("Please enter Easy, Medium, or Hard.")
```

```python

print("===== SMART STUDY QUIZ =====")

score = 0

answer = input("1. Python is a programming language? (yes/no): ")
if answer.lower() == "yes":
    print("Correct answer!")
    score += 1
else:
    print("Wrong answer!")

answer = input("2. Which language are we learning? ")
if answer.lower() == "python":
    print("Correct answer!")
    score += 1
else:
    print("Wrong answer!")

answer = input("3. How many minutes are there in one hour? ")
if answer == "60":
    print("Correct answer!")
    score += 1
else:
    print("Wrong answer!")

print("Your score:", score, "out of 3")
print("Keep learning every day!")
```
